```python
import pandas as pd
import geopandas as gpd
```

```python
paths_uri = 'movcud-zolaqlar-velo.kmz'
parkings_uri = 'mpv-v-mps.-son.kmz'

gpd.list_layers(paths_uri)
```

```python
paths_df = gpd.read_file(paths_uri, layer='Mövcud zolaqlar')
```

```python
paths_df.head()
paths_df_meter = paths_df.to_crs(epsg=32639)
```

```python
# Sual 1.1: Mövcud velosiped zolaqlarının (Mövcud zolaqlar qatı, EPSG:32639 proyeksiyası ilə) ümumi uzunluğu neçə kilometrdir (vergüldən sonra 2
# rəqəm) və ən çox velosiped zolağı olan küçə/prospekt hansıdır?
# (Format: "Uzunluq km, Küçə adı", məs: 17.19, 8 Səməd Vurğun k.)

total_lenght = 0.0
for path in paths_df_meter.length:
    total_lenght += path

length_total = round(total_lenght/1000,2)
```

```python
district_paths = paths_df_meter.groupby('Name').size().sort_values(ascending=False)
district_paths
```

```python
max_paths_district = district_paths.idxmax()
max_paths_district
```

```python
#1.1 CAVAB
print(f'{length_total}, {max_paths_district}')
```

```python
#2.	Sual 1.2: velosiped məntəqələri yeni qatındakı 185 velosiped parkinqindən neçəsi mövcud velosiped zolaqlarına 50 metr və daha yaxın məsafədə yerləşir?
# (Format: Yalnız tam ədəd daxil edin).

layers = gpd.list_layers(parkings_uri)
print(layers)
parkings_df = gpd.read_file(parkings_uri, layer ='velosiped məntəqələri yeni')
```

```python
points_meter = gpd.GeoSeries(parkings_df.geometry, crs = parkings_df.crs).to_crs(epsg=32639)
```

```python
relevant_p_count = 0
for point in points_meter:
    if (paths_df_meter.geometry.distance(point) <= 50.0).sum() > 0:
        relevant_p_count = relevant_p_count +1
```

```python
# CAVAB 1.2
print(relevant_p_count)
```

```python
#3.	Sual 1.3: Parkinq koordinatlarını araşdırdıqda Bakı mərkəzi şəhər çərçivəsindən (enlik 40.30–40.46, uzunluq 49.70–50.05) kənarda qalan neçə parkinq məntəqəsi var və adlarını qeyd edin?
# (Format: "Say: Məntəqə adları", məs: 4: MPV-144-2, …)
```

```python
arr = parkings_df[~((parkings_df.geometry.x > 49.70) & (parkings_df.geometry.x < 50.05) & (parkings_df.geometry.y > 40.30) & (parkings_df.geometry.y < 40.46))]
```

```python
# CAVAB 1.3
print(f'{len(arr)}: {", ".join(arr.Name)}')
```
